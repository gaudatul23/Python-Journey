```python
# function to clean text 
# convert lowercase 

def to_lowercase(text):
    return text.lower()
```

```python
df['cleaned_review'] = df['review'].apply(lambda x :to_lowercase(x))
```

```python
df
```

```python
# function to remove html tags
def remove_html_tags(text):
    clean = re.compile('<.*?>')
    return re.sub(clean, '', text)

df['cleaned_data']= df['cleaned_review'].apply(lambda x :remove_html_tags(x))
```

```python
# function to remove special characters and punctuation
def remove_special_characters(text):
    return re.sub(r'[^a-zA-Z\s]', '', text)
```

```python
df['cleaned_data'] = df['cleaned_data'].apply(lambda x :remove_special_characters(x))
```

```python
# Function to remove to numbers 
def remove_numbers(text):
    return re.sub(r'\d+', '', text)
```

```python
df['cleaned_data'] = df['cleaned_data'].apply(lambda x :remove_numbers(x))
```

```python
#function to remove extaspaces / whitespaces 
def remove_extra_spaces(text):
  return ''.join(text.split())

df['cleaned_data'] = df['cleaned_data'].apply(lambda x :remove_extra_spaces(x))
```

```python
# function to remove hastages 
def remove_hashtags(text):
    return re.sub(r'#\w+', '', text)

df['cleaned_data'] = df['cleaned_data'].apply(lambda x :remove_hashtags(x))
```
