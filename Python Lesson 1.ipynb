### Data Types ###

```python
# Data Types in Python 
# Python has several built-in data types, including:
# 1. Numeric Types: int, float, complex
# 2. Sequence Types: list, tuple, range
# 3. Text Type: str
# 4. Mapping Type: dict
# 5. Set Types: set, frozenset
# 6. Boolean Type: bool
# Examples of Data Types

# Numeric Types
age = 30  # int
height = 5.9  # float
complex_number = 2 + 3j  # complex

# Sequence Types
fruits = ["apple", "banana", "cherry"]  # list
coordinates = (10, 20)  # tuple
numbers = range(1, 10)  # range

# Text Type
name = "Atul"  # str

# Mapping Type
person = {"name": "Alice", "age": 25, "city": "New York"}  # dict

# Set Types
unique_numbers = {1, 2, 3, 4, 5}  # set

immutable_set = frozenset({1, 2, 3, 4, 5})  # frozenset

# Boolean Type
is_student = True  # bool

# Type Checking
# We can check the type of a variable using the type() function
print(type(age))  # Output: <class 'int'>
print(type(height))  # Output: <class 'float'>
print(type(name))  # Output: <class 'str'>
print(type(person))  # Output: <class 'dict'>
```

### Types of Conversion ###


```python
# Changing One data type into another data type is called Type Casting
# Type Casting
# We can change the type of a variable using type casting functions such as int(), float(), str(), etc.
# Example of Type Casting
age = 30
age_str = str(age)  # Converting int to str

height = 5.9
height_int = int(height)  # Converting float to int (this will truncate the decimal part)

number_str = "100"
number_int = int(number_str)  # Converting str to int
# Type casting can also be used to convert between other data types, such as lists, tuples, and sets.

fruits = ["apple", "banana", "cherry"]
fruits_tuple = tuple(fruits)  # Converting list to tuple

# this was and Explicit Type Casting because we are explicitly calling the type casting function to convert the data type.
```

```python
# Implicit Type Casting
# Implicit type casting is when Python automatically converts one data type to another without the need for explicit type casting functions.
# Example of Implicit Type Casting
x = 10  # int
y = 5.5  # float
result = x + y  # Python automatically converts x to float before performing the addition
print(result)  # Output: 15.5 (result is a float)
```

### Difference Between Implicit and Explicit ###

```python
# Difference between Explicit and Implicit Type Casting

# Explicit Type Casting:

# 1. The programmer manually converts one data type to another using type casting functions (e.g., int(), float(), str()).
# 2. It is done intentionally by the programmer to achieve a specific result.
# 3. It can lead to data loss if not done carefully (e.g., converting a float to an int will truncate the decimal part).

# Implicit Type Casting:
# 1. Python automatically converts one data type to another when necessary during operations.
# 2. It is done implicitly by Python without the programmer's intervention.
# 3. It is generally safe and does not lead to data loss, as Python handles the conversion appropriately.

```

### Input / Output Operation ###

```python
# Input Output in Python

# Input
# The input() function is used to take input from the user. It reads a line of text from the user and returns it as a string.
name = input("Enter your name: ")  # This will prompt the user to enter their name
age = input("Enter your age: ")  # This will prompt the user to enter their age

# Output
# The print() function is used to display output to the user. It can take multiple arguments and will print them separated by a space by default.
print("Hello, " + name + "! You are " + age + " years old.")  # This will print a greeting message with the user's name and age

# We can also use formatted strings (f-strings) for better readability
print(f"Hello, {name}! You are {age} years old.")  # This will achieve the same result using f-strings

# We can also print multiple variables in a single print statement
print("Name:", name, "Age:", age)  # This will print the name and age with labels

# We can also use the sep parameter to change the separator between arguments
print("Name:", name, "Age:", age, sep=" | ")  # This will print the name and age separated by a vertical bar

# We can also use the end parameter to change the end character of the print statement
print("Hello, " + name + "!", end=" ")  # This will print the greeting without a newline at the end

print("Welcome to Python programming!")  # This will continue on the same line as the previous print statement
```

```python
# Lets generate a Student Profile Project #
# Student Profile Project
name = (input("Enter your name:"))
age = (input("Enter your age:"))  # Converting the input to an integer
height = (input("Enter your height in meters:"))  # Converting the input to a float
city = input("Enter your city:") 
College = input("Enter your college name:")
Branch = input("Enter your branch:")
CGPA = (input("Enter your CGPA:"))  # Converting the input to a float
is_student = input("Are you a student? (yes/no):").lower() == "yes"  # Converting the input to a boolean

# Displaying the Student Profile
print("\nStudent Profile:")
print(f"Name: {name}")
print(f"Age: {age}")
print(f"Height: {height} meters")
print(f"City: {city}")
print(f"College: {College}")
print(f"Branch: {Branch}")
print(f"CGPA: {CGPA}")
print(f"Is Student: {'Yes' if is_student else 'No'}")

#here f represents formatted string literals, which allow us to embed expressions inside string literals using curly braces {}. 
# The expressions inside the curly braces are evaluated at runtime and their values are inserted into the string. 
# This makes it easier to create dynamic strings that include variable values or expressions without needing to concatenate strings manually.
```
