## String Properties ##

### Property 1 Ordered ###

```python
name = "Atul"

# here every characters carry an index starting from 0
print(name[0])  # Output: A
# A=0
# t=1
# u=2
# l=3
# python remembers the index of each character in the string,
#  so we can access them using their respective indices.
```

### Proerty 2 Indexed ###

```python
# Because of Order Exists The String is called a Sequence Data Type
# becuse of order exists every character has an index and we can access them using their respective indices.

name = "Atul Guad"
# Slicing
print(name[0:4])  # Output: Atul
print(name[0:9])

# now here the Middle Space between "Atul" and "Gaud" is also counted as a character and it has an index of 4

#here you can range the index as you want
# in square brackets you can specify the start index and end index and it will return the characters in that range
# here 0 means the start index and 4 means the end index but the end index is not included in the output

print(name[5:9])  # Output: Guad

a = "Hello     World   "
# here we have multiple spaces between Hello and World and also after World
print(a.strip())  # Output: Hello     World
# strip() method removes the leading and trailing whitespace from the string but it does not remove the whitespace between the words.
```

### Property 3 Immutable ###

Immutable means:

Once created, individual characters cannot be changed.

```python
name = "Python"

name[0] = "J"
# Output: TypeError: 'str' object does not support item assignment
# Strings are immutable, which means you cannot change individual characters in a string.
```

```python
name = "J" + name[1:]
print(name)  # Output: Jython
# here we are creating a new string by concatenating "J" with the substring of name starting from index 1 to the end of the string.
```

### Property 4 Iterable ###
Iterable means:

Python can go through each character one by one.

```python
for letter in "Python":
    print(letter)
    # It will print each letter of the string "Python" on a new line because we are iterating through each character in the string using a for loop.
    # The loop will run 6 times, once for each character in the string "Python", and it will print the current character in each iteration.
    
```

### Slicing ###

```python
# let have some negative indexing 
name = "Python"
print(name[-1])  # Output: n
print(name[-2])  # Output: o


email = "atul@gmail.com"
print(email.split("@"))  # Output: ['atul', 'gmail.com']
# here we are splitting the email string into a list of two parts using the "@" symbol as the separator. The split() method takes the separator as an argument and returns a list of substrings that were separated by the specified separator. In this case, it will return a list with two elements: 'atul' and 'gmail.com'.

print(email[-1])
# Output: m

#here we are accessing the last character of the email string using negative indexing. The index -1 refers to the last character in the string, which is 'm' in this case.

```

```python
name = "Atul Guad"
print(name[0:])
# here the 0: here we mentioned the start index not the end so , python internally grab the end index as the length of the string and return the characters from index 0 to the end of the string "Atul Guad".

print(name[:])
# here the : means that we have not specified the start index and end index, so it will return the entire string "Atul Guad".

print(name[:5])
#here the start indexed is not mentioned so , internally python grab the 0 index as a start and end to our given end index 
```

```python
# step Slicing 
name = "Atul Guad"
print(name[0:9:2])  

# see here [ 0 : 9 : 2 ] 
# 0 is the start index
# 9 is the end index but it is not included in the output
# 2 is the step which means we are taking every 2nd character from the range of index 0 to 9 (not included) and returning it as a new string "AlGa".

```

```python
xyz = "Hello World"
print(xyz.upper())  # Output: HELLO WORLD
print(xyz.lower())  # Output: hello world

print(xyz[::2])
# here we are using slicing with a step of 2 to get every second character from the string "Hello World". The syntax [::2] means that we are starting from the beginning of the string (index 0) and taking every second character until the end of the string. The output will be "HloWrd", which includes the characters at indices 0, 2, 4, 6, 8, and 10.
print(xyz[1::2])
# here we are using slicing with a step of 2 to get every second character from the string "Hello World", but this time we are starting from index 1 instead of index 0. The syntax [1::2] means that we are starting from index 1 and taking every second character until the end of the string. The output will be "el ol", which includes the characters at indices 1, 3, 5, 7, and 9.
print(xyz[::-1])


print(len(xyz))  # Output: 11
```

```python
# =====================================
# STRING METHODS IN PYTHON
# =====================================

name = "atul gaud"

# 1. upper()
print(name.upper())
# Output: ATUL GAUD


# 2. lower()
print(name.lower())
# Output: atul gaud


# 3. title()
print(name.title())
# Output: Atul Gaud


# 4. capitalize()
print(name.capitalize())
# Output: Atul gaud


# 5. swapcase()
print(name.swapcase())
# Output: ATUL GAUD


# =====================================
# MODIFYING METHODS
# =====================================

text = "I love Java"

# 6. replace()
print(text.replace("Java", "Python"))
# Output: I love Python


text = "   Python   "

# 7. strip()
print(text.strip())
# Output: Python


# 8. lstrip()
print(text.lstrip())
# Output: Python


# 9. rstrip()
print(text.rstrip())
# Output: Python


# =====================================
# SEARCH METHODS
# =====================================

word = "Programming"

# 10. find()
print(word.find("g"))
# Output: 3


# 11. count()
print(word.count("m"))
# Output: 2


# =====================================
# CHECKING METHODS
# =====================================

email = "atul@gmail.com"

# 12. startswith()
print(email.startswith("atul"))
# Output: True


# 13. endswith()
print(email.endswith(".com"))
# Output: True


# =====================================
# CONVERSION METHODS
# =====================================

skills = "Python,SQL,Excel"

# 14. split()
print(skills.split(","))
# Output: ['Python', 'SQL', 'Excel']


skills_list = ["Python", "SQL", "Excel"]

# 15. join()
print(" | ".join(skills_list))
# Output: Python | SQL | Excel


# =====================================
# VALIDATION METHODS
# =====================================

# 16. isalpha()
print("Atul".isalpha())
# Output: True


# 17. isdigit()
print("12345".isdigit())
# Output: True


# 18. isalnum()
print("Atul123".isalnum())
# Output: True


# 19. isspace()
print("   ".isspace())
# Output: True
```
