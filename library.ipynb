```python
# Match a word in a sentance
import re 

text = "I like to eat greenapple pie"

#search for apple in the text 
match = re.search(r"apple",text)     # here the r is for the raw string and the pattern is apple

#print the match object 
print(match.group()) # Output: apple
```

```python
# check if a string starts with hello 

text = "hello world"


# use ^ Carret to match the begining of the string 
match = re.match(r"hello",text)      # here the Carret is use for introgate sentence 

print(bool(match))
```

```python
# find all digit in a string 
text = "my phone number is 9876543210"
#\d match any digit , + means one or more times
digits = re.findall(r"\d+",text)
print(digits)

# / means whenever charac end and number start it consider number form there 
```

```python
# extract all words starting with a captial letter
text = "Alice and Bob are attending the Conference in New York"

# [A-Z] matches any capital letter , \w* matches the rest of the word
capital_words = re.findall(r"\b[A-Z]\w*",text)

print(capital_words)  # Output: ['Alice', 'Bob', 'Conference', 'New', 'York']
```

```python
# Replace all spaces with hyphens

text  = "This is a sample text"

# \s matches any space character
replaced = re.sub(r"\s"," =====> ",text)

print(replaced)  # Output: This-is-a-sample-text
```

```python
# match any 5 letter word
text = "These are some words: apple, table, chair, house, sky"

# \b indicates word boundary , \w{5} matches any 5 letter word
matches = re.findall(r"\b\w{5}\b",text)

print(matches)  # Output: ['apple', 'table', 'chair', 'house']
```

```python
# find words ending with 'ing'      real life example like eaml verifcication etc

text = "I am running and jumping while singing"

# \b indicates word boundary , ing matches words ending with 'ing'
ing_words = re.findall(r"\b\w+ing\b",text)     #here \w+ means one or more word character before ing
print(ing_words)  # Output: ['running', 'jumping', 'singing']
```

```python
# Extract date in format dd/mm/yyyy

text = "Today's date is 26/11/2025"
# \d{2} matches two digits , / is literal slash , \d{4} matches four digits
date = re.search(r"\d{2}/\d{2}/\d{4}",text)

print(date.group())  # Output: 26/11/2025
```

```python
# Match strings that contain only alphobets 
words = ["Hello", "World123", "Python3", "Regex", "Test!"]

# check each word for only alphabets using ^[A-Za-z]+$ pattern
for word in words:
    if re.match(r"^[A-Za-z]+$",word):
        print(word)

# Output: 
```

```python

email = "gaudatul@gmail.com"

# @ symbol followed by group of letter and . followed by group of letter
match = re.search(r"@([\w\.]+)",email)

print(match.group(1))  # Output: gmail.com
```

```python
# check if string is a valid indian phone number

number = "9876543210"

# start with 6-9 followed by 9 digits
pattern = r"^[6-9]\d{9}$"

print(bool(re.match(pattern,number)))  # Output: True 

#note when there is a word called name introgation and validation then the output is boolen 
```

```python
# Find repeated words in a sentance

text = "This is is a test test sentence"

# (\b\w+\b) captures a word , \s+ matches space , \1 refers to the first captured group
matches = re.findall(r"(\b\w+\b)\s+\1",text)

print(matches)  # Output: ['is', 'test']
```

```python
# Extract hashtags from a tweet
tweet = "Loving the new features in #Python3 and #Regex! #coding #Python3"

# followed by letters/digits/underscores
hashtags = re.findall(r"#(\w+)",tweet)

print(hashtags)  # Output: ['Python3', 'Regex', 'coding', 'Python3']
```

```python
# split a string by commas or semicolons

text = "apple,banana;cherry,dates;figs"
# split by , or ;
fruits = re.split(r"[;,]",text)
print(fruits)  # Output: ['apple', 'banana', 'cherry', 'dates', 'figs']
```
