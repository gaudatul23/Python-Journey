## Walrus Operator ##

```python
# Easy Walrus Operator Example

numbers = [1, 2, 3, 4, 5]
if (n := len(numbers)) > 3:
    print(f"The list has {n} elements, which is more than 3.")

# here the walrus operator (:=) is used to assign the length of the list to the variable n while also checking if it is greater than 3. This allows us to avoid calling len(numbers) twice and makes the code more concise.

```

## Ternary Operator ##

```python
# Understand the Ternary Operator in Python
# The ternary operator in python is a one-liner conditional statement that allows you to assign a value to a variable based on a condition. 
# The syntax for the ternary operator is as follows:
# variable = value_if_true if condition else value_if_false

# The ternary operator allows you to assign one value if a condition is true, and another if it is false:

# simple Example on Ternary Operator

age = 20
status = "Adult" if age >= 18 else "Minor"
print(status)  # Output: Adult

# Another Example on Ternary Operator
number = 10
result = "Even" if number % 2 == 0 else "Odd"
print(result)  # Output: Even

# deeper Example on Ternary Operator
score = 85
grade = "A" if score >= 90 else "B" if score >= 80 else "C" if score >= 70 else "D" if score >= 60 else "F"
print(grade)  # Output: B
```

```python
# Then whats the difference between the Ternary Operator and the if-else statement in Python?
# lets understand in simple way and detail format 

# The main difference between the ternary operator and the if-else statement in Python is that the
#  ternary operator is a more concise way to write simple conditional expressions, while the if-else statement is more suitable for complex conditions and multiple statements.

# Here are some key differences between the two:
# 1. Syntax: The ternary operator has a more compact syntax, while the if-else statement requires multiple lines of code.
# 2. Readability: The ternary operator can be less readable for complex conditions, while the if-else statement is generally more readable and easier to understand.
# 3. Use Cases: The ternary operator is best used for simple conditions and assignments, while the if-else statement is better suited for more complex logic and multiple statements.

# small example to illustrate the difference:
# Using the ternary operator
age = 20
status = "Adult" if age >= 18 else "Minor"
print(status)  # Output: Adult

# Using the if-else statement

age = 20
if age >= 18:
    status = "Adult"
else:
    status = "Minor"
print(status)  # Output: Adult

# In this example, the ternary operator provides a more concise way to assign the status based on the age, while the if-else statement requires more lines of code to achieve the same result.
```

## Comparison Operators ##


```python
# Lets see the Comparison Operators 
# Comparison operators in Python are used to compare two values and return a boolean result (True or False).
# Here is a list of comparison operators in Python:
# ==  Equal to
# !=  Not equal to
# >   Greater than
# <   Less than
# >=  Greater than or equal to
# <=  Less than or equal to
```

```python
# exmaple on comparison Operators 
x = 10
y = 5

print(x == y)  # Output: False
print(x != y)  # Output: True
print(x > y)   # Output: True
print(x < y)   # Output: False
print(x >= y)  # Output: True
print(x <= y)  # Output: False



```

```python
# another example on comparison Operators
Fruits = ["Apple", "Banana", "Cherry"]
print("Apple" in Fruits)  # Output: True
print("Mango" in Fruits)  # Output: False
```

```python
# Comparison Operators with Strings
str1 = "Hello"
str2 = "World"
print(str1 == str2)  # Output: False
print(str1 != str2)  # Output: True
print(str1 > str2)   # Output: False (because "Hello" comes before "World" in lexicographical order)
print(str1 < str2)   # Output: True (because "Hello" comes before "World" in lexicographical order)
```

## logical Operators ##

```python
# lets see the Logical Operators in Python
# Logical operators in Python are used to combine multiple conditions and return a boolean result (True or False).
# Here is a list of logical operators in Python:
# and  Logical AND operator
# or   Logical OR operator
# not  Logical NOT operator


```

```python
# Example of Logical Operators
x = 10
y = 5
print(x > 5 and y < 10)  # Output: True (both conditions are true)
print(x > 15 or y < 10)  # Output: True (one of the conditions is true)
print(not(x > 5))        # Output: False (the condition is true, so not operator returns false)

# Another Example of Logical Operators
age = 25
is_student = False
print(age > 18 and not is_student)  # Output: True (age is greater than 18 and is_student is False)


# Logical Operators with Strings
str1 = "Hello"
str2 = "World"
print(str1 == "Hello" and str2 == "World")  # Output: True (both conditions are true)
print(str1 == "Hello" or str2 == "Python")  # Output: True (one of the conditions is true)
print(not(str1 == "Hello"))  # Output: False (the condition is true, so not operator returns false)
```

## Identity Operator ##

Identity operators are used to compare the objects, not if they are equal, but if they are actually the same object, with the same memory location:

is operator : Returns True if both variables are the same object 
Example X is Y 

is not Operator : Returns True if both variables are not the same object
Example X is not Y 

```python
# lets see the Is operator in Python
# The is operator in Python is used to compare the identity of two objects. It checks whether two variables point to the same object in memory, rather than comparing their values.
# Here is an example of the is operator:
a = [1, 2, 3]
b = a  # b points to the same list as a
c = [1, 2, 3]  # c is a different list with the same content
print(a is b)  # Output: True (a and b point to the same object)
print(a is c)  # Output: False (a and c are different objects, even though they have the same content)

# Another Example of the is operator  
x = 10
y = 10
print(x is y)  # Output: True (small integers are cached by Python, so x and y point to the same object)
# However, for larger integers, Python may create separate objects:

x = 1000
y = 1000
print(x is y)  # Output: False (x and y may point to different objects in memory)


```

```python
x = ["apple", "banana"]
y = ["apple", "banana"]
z = x

print(x is z)
print(x is y) # here x and y are different objects in memory, even though they have the same content, so x is y will return False.
print(x == y) # now they are equal because they have the same content, so x == y will return True.
```

```python
# lets see the IS NOT operator in Python
# The is not operator in Python is used to compare the identity of two objects. It checks whether two variables do not point to the same object in memory.
# Here is an example of the is not operator:

a = [1, 2, 3]
b = a  # b points to the same list as a
c = [1, 2, 3]  # c is a different list with the same content
print(a is not b)  # Output: False (a and b point to the same object)
print(a is not c)  # Output: True (a and c are different objects, even though they have the same content)

# Another Example of the is not operator
x = 10
y = 10
print(x is not y)  # Output: False (small integers are cached by Python, so x and y point to the same object)
# However, for larger integers, Python may create separate objects:
x = 1000
y = 1000
print(x is not y)  # Output: True (x and y may point to different objects in memory)

x = ["apple", "banana"]
y = ["apple", "banana"]

print(x is not y) # here x and y are different objects in memory, even though they have the same content, so x is not y will return True.
```

```python
# difference between is and == operator in Python 
# lets see in simple way and detail format small lines 
# The is operator checks for identity, meaning it checks whether two variables point to the same object in memory.
# The == operator checks for equality, meaning it checks whether the values of the variables are equal, regardless of whether they point to the same object in memory.

# Example to illustrate the difference between is and == operator
a = [1, 2, 3]
b = a  # b points to the same list as a
c = [1, 2, 3]  # c is a different list with the same content
print(a is b)  # Output: True (a and b point to the same object)
print(a is c)  # Output: False (a and c are different objects, even though they have the same content)
print(a == c)  # Output: True (a and c have the same content, so they are considered equal)
```

## Membership Operators ##
Membership operators are used to test if a sequence is presented in an object:

in operator : Returns True if a sequence with the specified value is present in the object 
example X in Y 

not in operator : Returns True if a sequence with the specified value is not present in the object 
Example x not in Y 

```python
# lets see the In operator in Python
# The in operator in Python is used to check if a value exists within an iterable (like a list, tuple, string, etc.). It returns True if the value is found and False otherwise.

# Here is an example of the in operator:

fruits = ["apple", "banana", "cherry"]
print("banana" in fruits)  # Output: True (because "banana" is in the list)
print("grape" in fruits)   # Output: False (because "grape" is not in the list)

# Another Example of the in operator

text = "Hello, world!"
print("world" in text)  # Output: True (because "world" is in the string)
print("Python" in text)  # Output: False (because "Python" is not in the string)

```

```python
# lets see the Not in operator in Python
# The not in operator in Python is used to check if a value does not exist within an iterable (like a list, tuple, string, etc.). It returns True if the value is not found and False otherwise.

# Here is an example of the not in operator:

fruits = ["apple", "banana", "cherry"]
print("grape" not in fruits)  # Output: True (because "grape" is not in the list)
print("banana" not in fruits)  # Output: False (because "banana" is in the list)

# Another Example of the not in operator

text = "Hello, world!"
print("Python" not in text)  # Output: True (because "Python" is not in the string)
print("world" not in text)   # Output: False (because "world" is in the string)
```

## Operator Precedence ##
Operator precedence describes the order in which operations are performed.


```python
# lets get to Operator prdecence in Python
# Operator precedence in Python determines the order in which operators are evaluated in an expression.
# Here is a list of operator precedence in Python, from highest to lowest:
# 1. Parentheses ()
# 2. Exponentiation (**)
# 3. Unary plus and minus (+x, -x)
# 4. Multiplication, Division, Floor Division, Modulus (*, /, //, %)
# 5. Addition and Subtraction (+, -)
# 6. Bitwise Shift Operators (<<, >>)
# 7. Bitwise AND (&)
# 8. Bitwise XOR (^)
# 9. Bitwise OR (|)
# 10. Comparison Operators (==, !=, >, <, >=, <=)
# 11. Logical NOT (not)
# 12. Logical AND (and)
# 13. Logical OR (or)
# 14. Assignment Operators (=, +=, -=, *=, /=, etc.)
```

```python
# Left to Right Associativity
result = 10 - 5 + 2
print(result)  # Output: 7 (because subtraction and addition have the same precedence, so they are evaluated from left to right)
```
