## Error Handling 

```python
try:
  a = 10 # Example numerator
  b = 0  # Example denominator that causes division by zero
  result = a/b  # This will raise a ZeroDivisionError 

except ZeroDivisionError:
  print("Cannot divide by zero") # Handle division by zero error
```

## Handle Value Error 

```python
try:
  num = int("abc") # This will raise a ValueError

except ValueError: # This will raise a ValueError
  print("Invalid input: not an integer") # Handle invalid integer conversion
```

## 3. Handle Multiple exceptions

```python
try:
  value = 10/0
except (ZeroDivisionError, TypeError):
  print("an error occurred")
```

## 4. Use else block

```python
try:
  num =int("23")

except ValueError:
  print("Invalid input: not an integer")
else:
  print("Conversion successful:", num)

```

## 5. Use Finally block

```python
try:
  print("Opening File....")
  x = 10/2
except Exception:
  print("Error Occured")
finally:
  print(x,"Execution Completed")
```

## 6. Custom exception class 

```python
class Ageerror(Exception):
    pass
try:
  age = -5
  if age < 0:
      raise Ageerror("Age cannot be negative")

except Ageerror as e:
  print(e)
```

## 7. Raise built - in exception

```python
try:
  salary = -1000
  if salary < 0:
      raise ValueError("Salary cannot be negative")
except ValueError as e:
  print(e)
```
